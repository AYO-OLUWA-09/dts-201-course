# ***Class 3: Data collection and data sampling***

```python
import pandas as pd

df = pd.read_excel('/50_startups_messy (1).xlsx')

print("Shape: ", df.shape)

print("Missing values: ", df.isnull().sum())

print("Duplicate rows: ", df.duplicated().sum())
```

```python
##Drop duplicate: there is no duplicate values so we just move on

#standardize state
df['State'] = df['State'].str.strip().str.title()

## convert maerketing spend to numeric

df['Marketing Spend'] = pd.to_numeric(df['Marketing Spend'], errors='coerce')
```

```python

# simple random
srs = df.sample(n=20, random_state=7)


# stratified, 20% of each state
strat = df.groupby("State", group_keys=False ) \
          .apply(
lambda g: g.sample(frac=0.2))
```

```python
# View the R&D Spend and Marketing Spend for the last 4 startups

last_4 = df[['R&D Spend', 'Marketing Spend']].tail(4)
print(last_4)

# Compare the averages of the last 4 startups
last_4_means = last_4.mean()
print("\nAverages for last 4 startups:")
print(last_4_means)
```

```python
# Create the comparison table
compare = pd.DataFrame({
    "Population": df[['R&D Spend', 'Marketing Spend']].mean(),
    "SRS (20)": srs[['R&D Spend', 'Marketing Spend']].mean(),
    "Stratified (20%)": strat[['R&D Spend', 'Marketing Spend']].mean()
})

# Display the comparison rounded to 2 decimal places
print(compare.round(2))
```
